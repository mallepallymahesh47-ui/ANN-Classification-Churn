```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

```python
# Load the trained Model
# Load scalarz pickle and onehot

model=load_model("model.h5")

# Load the encoder and scalar
with open("onehot_encoder_geo.pkl", "rb") as f:
    label_encoder_geo=pickle.load(f) 

with open("encoder_gender.pkl", "rb") as f:
    label_encoder_gender=pickle.load(f) 

with open("scaler.pkl", "rb") as f:
    scaler=pickle.load(f) 
```

```python
# Example Data
input_data={
    "CreditScore":600,
    "Geography":"France",
    "Gender":"Male",
    "Age":23,
    "Tenure":1,
    "Balance":60000,
    "NumOfProducts":2,
    "HasCrCard":1,
    "IsActiveMember":1,
    "EstimatedSalary":50000,
}
```

```python
encoder_geo=label_encoder_geo.transform([[input_data["Geography"]]])
encoded_geo=pd.DataFrame(encoder_geo.toarray(), columns=label_encoder_geo.get_feature_names_out(['Geography']))
encoded_geo
```

```python
input_df=pd.DataFrame([input_data])
input_df
```

```python
# Encode Categorical Variables
input_df["Gender"]=label_encoder_gender.transform(input_df["Gender"])
input_df
```

```python
# Concatination one hot Encoded

input_df=pd.concat([input_df.drop("Geography", axis=1), encoded_geo], axis=1)
input_df
```

```python
# Scaling the input 
input_scaled=scaler.transform(input_df)
input_scaled
```

```python
# Prediction Churn
Prediction=model.predict(input_scaled)
Prediction
```

```python
Prediction_prop=Prediction[0][0]
Prediction_prop
```

```python
if Prediction_prop>0.5:
    print("The Customer is likely to churn")
else:
    print("The Customer is not likely to churn")
```
